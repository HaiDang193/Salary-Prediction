```python
import pandas as pd
from pathlib import Path

# 1. Xác định thư mục data
ROOT = Path.cwd()
DATA_DIR = ROOT/ ".." / "data" / "raw"

# 2. Đọc dữ liệu JSON
JSON_PATH = DATA_DIR / "du_lieu_adzuna_tho.json"

df = pd.read_json(JSON_PATH)

# 3. Chuyển JSON sang CSV
CSV_PATH = DATA_DIR / "du_lieu_adzuna_tho.csv"

df.to_csv(
    CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)
```
