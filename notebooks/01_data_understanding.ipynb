Salary Unit Assumption (Giả định về đơn vị lương)

Dự án xây dựng mô hình dự đoán trung điểm khoảng lương (salary_mid) dựa trên dữ liệu tuyển dụng tại Mỹ được thu thập thông qua Adzuna API.

Do dataset không chứa trường xác định kỳ trả lương riêng cho từng bản ghi, nghiên cứu sử dụng giả định rằng phần lớn các giá trị lương được biểu diễn theo cơ sở năm (USD/year). Giả định này phù hợp với đặc điểm phân phối lương quan sát được nhưng chưa được xác minh độc lập cho toàn bộ dữ liệu.

Trong quá trình kiểm tra, 12 bản ghi có giá trị lương dưới ngưỡng rà soát được tạm loại khỏi tập dữ liệu baseline do có nguy cơ không nhất quán về đơn vị hoặc chất lượng giá trị.

Ngưỡng $10,000 chỉ đóng vai trò hỗ trợ phát hiện các trường hợp đáng nghi, không được xem là bằng chứng xác nhận kỳ trả lương của các bản ghi còn lại.

Vì vậy, kết quả mô hình được diễn giải là mức lương trung điểm dự đoán theo giả định lương năm. Việc thiếu thông tin kỳ trả lương là một hạn chế về chất lượng nhãn của nghiên cứu.

```python
import pandas as pd
from pathlib import Path

# 1. Xác định thư mục data
ROOT = Path.cwd()
DATA_DIR = ROOT/ ".." / "data" / "raw"

# 2. Đọc dữ liệu JSON
JSON_PATH = DATA_DIR / "du_lieu_adzuna_tho.json"

df = pd.read_json(JSON_PATH)

# 3. Chuyển JSON sang CSV
CSV_PATH = DATA_DIR / "du_lieu_adzuna_tho.csv"

df.to_csv(
    CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)
```

Giải thích các cột trong file csv:
1. Dataset Description (Mô tả dữ liệu gốc)
Dữ liệu được thu thập từ Adzuna API, bao gồm thông tin các tin tuyển dụng tại Mỹ như vị trí công việc, doanh nghiệp, địa điểm, ngành nghề, mức lương và thời điểm đăng tuyển.

Dataset ban đầu gồm 2.403 bản ghi và 17 thuộc tính.
1.1. Data Dictionary

- id : Mã định danh của tin tuyển dụng trên Adzuna, được sử dụng để kiểm tra và xử lý các bản ghi trùng lặp.
- title : Tên vị trí tuyển dụng (Job Title), ví dụ: Software Engineer, Data Analyst.
- description : Nội dung mô tả công việc, bao gồm các thông tin được cung cấp trong tin tuyển dụng.
- company : Thông tin công ty tuyển dụng, được lưu dưới dạng cấu trúc dictionary.
- category : Thông tin ngành nghề hoặc nhóm công việc, được lưu dưới dạng dictionary.
- location : Thông tin địa điểm tuyển dụng, bao gồm tên địa điểm và các cấp khu vực địa lý.
- latitude : Vĩ độ của địa điểm tuyển dụng.
- longitude : Kinh độ của địa điểm tuyển dụng.
- contract_type : Loại hợp đồng lao động, ví dụ: permanent (hợp đồng lâu dài).
- contract_time : Hình thức thời gian làm việc, ví dụ: full_time (toàn thời gian).
- salary_min : Giá trị mức lương tối thiểu được Adzuna cung cấp cho tin tuyển dụng.
- salary_max : Giá trị mức lương tối đa được Adzuna cung cấp cho tin tuyển dụng.
- salary_is_predicted : Cờ xác định mức lương có được Adzuna đánh dấu là ước tính hay không. Giá trị 1 là có ước tính, 0 là không được đánh dấu là ước tính.
- created : Thời điểm tin tuyển dụng được tạo hoặc đăng trên hệ thống, ở định dạng thời gian ISO 8601.
- redirect_url : Đường dẫn tới trang chi tiết của tin tuyển dụng.
- adref : Mã tham chiếu do Adzuna cung cấp cho tin tuyển dụng.(không quan trọng)
- __CLASS__ : Metadata thể hiện loại đối tượng được trả về từ Adzuna API.(không quan trọng)
