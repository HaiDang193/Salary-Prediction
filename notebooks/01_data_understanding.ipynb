Tổng quan dự án
Dự án này phân tích một tập dữ liệu chứa thông tin chi tiết về các xe ô tô đã qua sử dụng được niêm yết trên CarDekho . Mục tiêu là khám phá các mô hình định giá, mức độ phổ biến của thương hiệu, sở thích về loại nhiên liệu và các thông tin thị trường khác để hỗ trợ người mua, người bán và các đại lý ô tô trong việc ra quyết định.

Mục tiêu:
Hiểu rõ sự phân bố giá xe dựa trên thương hiệu, mẫu mã và năm sản xuất.
Phân tích loại nhiên liệu và sở thích về hộp số.
So sánh sự khác biệt về giá dựa trên tuổi đời và số km đã đi của xe.
Xác định các thương hiệu phổ biến và thị phần của chúng.

Bộ dữ liệu : CarDekho trên kaggle
Loại tệp: CSV
Các cột chính:

name– Tên xe kèm nhãn hiệu và mẫu xe
year– Năm sản xuất
selling_price– Giá xe (INR)
km_driven– Tổng số km đã đi
fuel– Loại nhiên liệu (Xăng/Dầu diesel/Khí CNG/Khí LPG/Điện)
seller_type– Cá nhân/Đại lý/Đại lý Trustmark
transmission– Thủ công/Tự động
owner– Tình trạng sở hữu
